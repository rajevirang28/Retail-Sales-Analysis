```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../Dataset/retail_sales.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
# Total Revenue
total_revenue = df["Total Sales"].sum()
print(f"Total Revenue : ${total_revenue:,.2f}")
```

```python
# Total Orders
total_orders = df["Invoice"].nunique()
print("Total Orders:", total_orders)
```

```python
# Total Customers
total_customers = df["Customer ID"].nunique()
print("Total Customers:", total_customers)
```

```python
# Total Products
total_products = df["Description"].nunique()
print("Total Products:", total_products)
```

```python
# Average Order Value
avg_order_value = total_revenue / total_orders
print(f"Average Order Value:  ${avg_order_value:,.2f}")
```

```python
# Average Quantity Sold
avg_quantity_sold = df["Quantity"].mean()
print(f"Average Quantity Sold: ${avg_quantity_sold:.2f}")
```

```python
# Monthly Sales Trend
monthly_sales = (
    df.groupby(["Year", "Month Number", "Month"])["Total Sales"].sum().reset_index()
)

monthly_sales = monthly_sales.sort_values(
    by=["Year", "Month Number"]
)

monthly_sales
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    monthly_sales['Month'],
    monthly_sales["Total Sales"],
    marker="o"
)
plt.xticks(rotation=45)
plt.title("Monthly Slaes Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)
plt.show()
```

```python
# Top 10 Customers
top_customers = (
    df.groupby("Customer ID")["Total Sales"].sum().sort_values(ascending=False).head(10)
)
top_customers
```

```python
plt.figure(figsize=(12,6))
top_customers.sort_values().plot(kind="barh")
plt.title("Top 10 Customers")
plt.xlabel("total_revenue")
plt.show()
```

```python
# Top 10 Products
top_products = (
    df.groupby("Description")["Total Sales"].sum().sort_values(ascending=False).head(10)
)
top_products
```

```python
plt.figure(figsize=(12,6))
top_products.sort_values().plot(kind="barh")
plt.title("Top 10 Products")
plt.xlabel("Revenue")
plt.show()
```

```python
# Sales by Country

country_sales = (
    df.groupby("Country")["Total Sales"].sum().sort_values(ascending=False)
)
country_sales.head(10)
```

```python
plt.figure(figsize=(12,6))
country_sales.head(10).sort_values().plot(kind="barh")
plt.title("Top Countries by Sales")
plt.xlabel("Revenue")
plt.show()
```

```python
# Best Selling Weekday
weekday_sales = (
    df.groupby("Weekday")["Total Sales"].sum()
)

weekday_sales
```

```python
weekday_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]
weekday_sales = weekday_sales.reindex(weekday_order)

plt.figure(figsize=(12,6))
weekday_sales.plot(kind="bar")
plt.title("Sales by Weekday")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
# Busiest Shopping Hour
hourly_sales = (
    df.groupby("Hour")["Total Sales"].sum()
)

hourly_sales
```

```python
plt.figure(figsize=(12,6))

hourly_sales.plot(marker="o")

plt.title("Revenue by Hour")

plt.xlabel("Hour")

plt.ylabel("Revenue")

plt.grid(True)

plt.show()
```
